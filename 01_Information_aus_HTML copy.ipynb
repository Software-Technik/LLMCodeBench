```python
import requests
import os

# Advent of Code session cookie
# session_cookie = 'YOUR_SESSION_COOKIE_HERE'
cookies = {'session': session_cookie}

# Base URL for accessing the input
base_url = 'https://adventofcode.com/{year}/day/{day}/input'

# Project root directory where the files will be saved
project_root = 'project_root'

# Loop over years (from 2015 to 2024, represented as 15 to 24)
for year in range(14, 25):
    full_year = 2000 + year  # Convert the year to 20XX format
    year_url = base_url.format(year=full_year, day=1)  # Check the URL for the first day to verify the year

    # Make an initial request to verify if the year exists
    response = requests.get(year_url, cookies=cookies)

    # If the year doesn't exist (status code 404 or login required), skip it
    if response.status_code == 404 or "Please log in" in response.text or (year >= 2025 and response.status_code == 404):
        print(f"Year {full_year} does not exist or requires login. Skipping.")
        continue

    print(f'Year {full_year}:')

    saved_days = []  # List to keep track of saved days
    failed_days = []  # List to keep track of failed days

    for day in range(1, 32):  # Loop over days 1 to 31
        url = base_url.format(year=full_year, day=day)  # Format the URL with the current year and day
        response = requests.get(url, cookies=cookies)  # Make the request with the session cookie

        # Check if login is required or the request was unsuccessful
        if "Please log in" in response.text or response.status_code != 200:
            failed_days.append(day)  # Mark day as failed if login or status code is invalid
            continue

        # Create the target folder path for saving the input
        folder_path = os.path.join(project_root, str(full_year), f'challenge{day:02d}')
        os.makedirs(folder_path, exist_ok=True)  # Ensure the directory exists (creates it if not)

        # Define the file path for the input file
        file_path = os.path.join(folder_path, 'inputs')

        # Save the content to the file (strip leading/trailing whitespace)
        with open(file_path, 'w', encoding='utf-8') as f:
            f.write(response.text.strip())

        saved_days.append(day)  # Mark day as successfully saved

    # Print summary for the year
    saved_count = len(saved_days)
    failed_count = len(failed_days)
    print(f'  Days {saved_count} saved, Days {failed_count} not saved.')
    if failed_count > 0:
        print(f'  Failed days: {", ".join(map(str, failed_days))}')
```
